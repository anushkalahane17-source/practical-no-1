```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("retail_sales_dataset (1) (1).csv")

# Display First 5 Rows
print(df.head())

# Dataset Information
print(df.info())

# Statistical Summary
print(df.describe())
```

**1. Bar Plot**

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Gender', data=df)
plt.title("Gender Distribution")
plt.show()
```

**2. Pie Chart**

```python
gender = df['Gender'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender,
        labels=gender.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Gender Distribution")
plt.show()
```

**3 Histogram**

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=10, kde=True)
plt.title("Age Distribution")
plt.show()
```

**4. KDE Plot**

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Total Amount'], fill=True)
plt.title("KDE Plot of Total Amount")
plt.show()
```

5. Box **Plot**

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Total Amount'])
plt.title("Box Plot of Total Amount")
plt.show()
```

**6. Violin Plot**

```python
plt.figure(figsize=(8,5))
sns.violinplot(x='Gender', y='Total Amount', data=df)
plt.title("Violin Plot")
plt.show()
```

**7. Scatter Plot**

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Age', y='Total Amount', data=df)
plt.title("Age vs Total Amount")
plt.show()
```

8. Line **Plot**

```python
plt.figure(figsize=(10,5))
sns.lineplot(x='Age', y='Total Amount', data=df)
plt.title("Line Plot")
plt.show()
```

**9. Count Plot**

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Product Category', data=df)
plt.xticks(rotation=45)
plt.title("Product Category Count")
plt.show()
```

**10. Heatmap**

```python
plt.figure(figsize=(8,5))
corr = df.select_dtypes(include='number').corr()

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

**11. Pair Plot**

```python
sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

**12. Strip Plot**

```python
plt.figure(figsize=(8,5))
sns.stripplot(x='Gender', y='Total Amount', data=df)
plt.title("Strip Plot")
plt.show()
```

**13. Swarm Plot**

```python
plt.figure(figsize=(8,5))
sns.swarmplot(x='Gender', y='Total Amount', data=df)
plt.title("Swarm Plot")
plt.show()
```
